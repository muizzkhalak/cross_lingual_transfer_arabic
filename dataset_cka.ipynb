# MADAR

```python
import pandas as pd
import numpy as np
import os

madar_dialect_list = os.listdir('data/CKA/MADAR_Corpus/')
madar_dialect_list = [i for i in madar_dialect_list if '.tsv' in i and 'MADAR.corpus.' in i and 'ENG' not in i and 'index' not in i]

aleppo_reference = pd.read_csv('data/CKA/MADAR_Corpus/MADAR.corpus.Aleppo.tsv', delimiter = '\t')

if not os.path.exists('data_cleaned/CKA/MADAR_Corpus/'):
    os.makedirs('data_cleaned/CKA/MADAR_Corpus/', exist_ok=True)

for i in madar_dialect_list:
    df = pd.read_csv('data/CKA/MADAR_Corpus/' + i, delimiter = '\t')
    df = df[df['sentID.BTEC'].isin(aleppo_reference['sentID.BTEC'].tolist())]

    with open(f'data_cleaned/CKA/MADAR_Corpus/MADAR_{i.split(".")[2]}.txt', 'w') as file:
        for sent in df['sent'].tolist():
            file.write(f"{sent}\n")
```
